```python
import pandas as pd
import numpy as np
import pickle as pk

import matplotlib.pyplot as plt

from sklearn.metrics import average_precision_score, roc_auc_score
```

```python
y_cv = pk.load(open("../../checkpoints/cross_validation_test.pk", "rb"))
y_cv
```

```python
y_cv_no_conservation = pk.load(open("./ablation_cv/cv_predictions_no_conservation.pk", "rb"))
y_cv_no_conservation
```

```python
y_cv_no_position = pk.load(open("./ablation_cv/cv_predictions_no_position.pk", "rb"))
y_cv_no_position
```

```python
y_cv_no_mlc = pk.load(open("./ablation_cv/cv_predictions_no_mlc.pk", "rb"))
y_cv_no_mlc
```

```python
y_cv_no_ddg = pk.load(open("./ablation_cv/cv_predictions_no_ddG.pk", "rb"))
y_cv_no_ddg
```

```python
y_cv_no_rnamsm = pk.load(open("./ablation_cv/cv_predictions_no_rnamsm.pk", "rb"))
y_cv_no_rnamsm
```

```python
cv_AUC = pd.DataFrame({
    key: np.array([roc_auc_score(y_true, y_pred) for y_true, y_pred in zip(y_cv_i[0], y_cv_i[1])])
    for y_cv_i, key in zip(
        [y_cv, y_cv_no_position, y_cv_no_ddg, y_cv_no_mlc, y_cv_no_rnamsm, y_cv_no_conservation],
        ["full_model", "no_position", "no_ddg", "no_mlc", "no_rnamsm", "no_conservation"]
    )
})
cv_AUC
```

```python
cv_AP = pd.DataFrame({
    key: np.array([average_precision_score(y_true, y_pred) for y_true, y_pred in zip(y_cv_i[0], y_cv_i[1])])
    for y_cv_i, key in zip(
        [y_cv, y_cv_no_position, y_cv_no_ddg, y_cv_no_mlc, y_cv_no_rnamsm, y_cv_no_conservation],
        ["full_model", "no_position", "no_ddg", "no_mlc", "no_rnamsm", "no_conservation"]
    )
})
cv_AP
```

```python
AUC_drop = (cv_AUC.T - cv_AUC.full_model).T.drop("full_model", axis=1).apply(
    lambda col: [col.sample(len(col), replace=True).mean() for _ in range(10_000)]
)
AUC_drop
```

```python
AP_drop = (cv_AP.T-cv_AP.full_model).T.drop("full_model", axis=1).apply(
    lambda col: [col.sample(len(col), replace=True).mean() for _ in range(10_000)]
)
AP_drop
```
